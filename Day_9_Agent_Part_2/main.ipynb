```python
REVIEWS = {
    "how to train your dragon": {
        "reviews": 1520,
        "rating": 4.8
    },

    "harry potter and the philosophers stone": {
        "reviews": 3240,
        "rating": 4.8
    },

    "the lion king": {
        "reviews": 2180,
        "rating": 4.9
    },

    "the chronicles of narnia": {
        "reviews": 1890,
        "rating": 4.8
    },

    "the hobbit": {
        "reviews": 2760,
        "rating": 4.9
    },

    "percy jackson and the lightning thief": {
        "reviews": 1980,
        "rating": 4.8
    },

    "peter pan": {
        "reviews": 1430,
        "rating": 4.7
    },

    "the jungle book": {
        "reviews": 1650,
        "rating": 4.7
    }
}
```

```python
@tool
def get_book(name: str) -> str:
    """Look up a book by name and return its price, rating, author, and description."""
    b = BOOKS.get(name.lower())
    if not b:
        return f"Book not found. Available: {', '.join(BOOKS)}"
    return str(b)

@tool
def get_review(name: str) -> str:
    """Look up reviews for a book by name and return the number of reviews and rating."""
    r = REVIEWS.get(name.lower())
    if not r:
        return f"Review not available for this book"
    return str(r)
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    groq_api_key=API_KEY
)
```

```python
agent2 = create_agent(
    llm,
    tools=[get_book, get_review],
    system_prompt="You are a helpful book assistant for a family-friendly online bookstore.",
    checkpointer=InMemorySaver()
)
```

```python
def ask2(question: str):
    config = {
        "configurable": {
            "thread_id": "user-alice-session-1"
        }
    }

    result = agent2.invoke(
        {
            "messages": [
                {
                    "role": "user",
                    "content": question
                }
            ]
        },
        config=config
    )

    print(result["messages"][-1].content)
```

```python
ask2("Which book is about a magical school?")
```

```python
ask2("What about its reviews?")
```
